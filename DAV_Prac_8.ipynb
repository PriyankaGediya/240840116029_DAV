Practical 8:Create interactive visualizations using Plotly Express and Plotly Graph Objects.


Step 1: Import requried library

```python
# Import Pandas for data manipulation
import pandas as pd

# Import plotly express for quick and easy visualization
import plotly.express as px

#Import plotly graph objects for customized visualization
import plotly.graph_objects as go
```

Step 2: Create a sample data

```python
Data = {
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    "Product": ["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet"],
    "Region": ["North","South","East","West","North","South","East","West","North","South","East","West"],
    "Sales": [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Units_Sold": [12,30,25,14,35,30,18,38,32,21,45,40],
    "Profit": [18000,27000,30000,22000,32000,35000,28000,34000,38000,33000,46000,50000]
}
```

```python
#Convert the dictionary into a Pandas DataFrame.
df = pd.DataFrame(Data)
df
```

```python
# Display the first five records
df.head()
```

```python
# Display the number of row and columns
df.shape
```

```python
#Display the column names and data types.
df.info()
```

Step 3:Create an Interactive line chart using plotly express

```python
#Sales revenu change the throughout the year
figure = px.line(df, x="Month",y="Sales",title="Sales revenue through the year")
figure.show()
```

Step 4: Improve the hover information

```python
#Create a line chart using Sales over Month
#Add the additional columns to the hover information. Students cam see product, region , unit sold and profit when they move the mouse over a data point
figure = px.line(df, x="Month",y="Sales",color="Product",title="Sales revenue through the year")
figure.show()
```

Step 5: Create an Interactive Bar Chart

```python
#Group the dataset according to the product
figure = px.bar(df, x="Month",y="Units_Sold",color="Product",title="Units sold per month")
figure.show()
```

```python
#Calculate the total sales generated by each product
figure = px.bar(df, x="Product",y="Sales",color="Region",title="Total sales generated by each product")
figure.show()
```

```python
#Create an interactive bar chart
figure = px.bar(df, x="Product",y="Sales",color="Region",title="Total sales generated by each product")
figure.show()
```

Step 6: Interactive Scatter Plot

```python
#To analyze the realtionship between units sold and sales using an interactive Scatter Plot.
figure = px.scatter(df, x="Units_Sold",y="Sales",size="Profit",color="Product",title="Relationship between Units Sold and Sales")
figure.show()
```

Step 7: Interactive Scatter Plot with color

```python
# we will use region as the color variable
figure = px.scatter(df, x="Units_Sold",y="Sales",size="Profit",color="Region",title="Relationship between Units Sold and Sales")
figure.show()
```

Step 8: Create an Interactive Pie chart

```python
# To visualize regional sales contribution using an interactive pie chart
figure = px.pie(df, values="Sales",names="Region",title="Regional Sales Contribution")
figure.show()
```

Step 9: Plotly Graph Objects

```python
# To create an interactive visualization using plotly graph objects and understand how it differs from plotly express
figure = go.Figure(data=[go.Scatter(x=df.Month,y=df.Sales,mode='lines+markers')])
figure.show()
```

Step 10: Create line chart using graph objects

```python
#Create a Figure object
figure = go.Figure()
```

```python
# Add a line trace to the Figure. This trace represents the monthly sales data
figure.add_trace(go.Scatter(x=df.Month,y=df.Sales,mode='lines+markers',name='Sales'))
```

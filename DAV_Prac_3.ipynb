```python
#Pandas is used to create, read, modify, and analyze tabular data.
import pandas as pd
```

```python
#add data
data = {
    "Emp_ID":[101,102,103,104,105,106,107,108],
    "Name":["Rahul","Priya","Amit","Riya","Karan","Neha","Vivek","Pooja"],
    "Department":["IT","HR","IT","Finance","HR","Finance","IT","HR"],
    "City":["Surat","Navsari","Surat","Valsad","Surat","Navsari","Valsad","Surat"],
    "Salary":[45000,38000,52000,48000,40000,51000,60000,42000],
    "Experience":[2,1,5,4,2,6,8,3]
}
```

Step 1 : Display the Dataset

```python
#Convert the dictionary into a Pandas DataFrame for easier data manipulation
df = pd.DataFrame(data)
df
```

Step 2 : Filter Data

```python
#Select only employees working is the IT department
df[df["Department"]=="IT"]
```

Display Employees with salary greater than 50,000

```python
#Display the employees whos salary is greater than 50000
df[df["Salary"]>50000]
```

Display employees from Surat

```python
#Display employees from Surat
df[df["City"]=="Surat"]
```

Apply Multiple Conditions

```python
#display employee from the IT department
#having salary greater than 50000
df[(df["Department"]=="IT") & (df["Salary"]>50000)]
```

Step 3 : Sort Data

Sort Salary in Ascending Order

```python
#Sort employee by salary from lowest to highest.
df.sort_values(by="Salary")
```

Sort salary in Descending Order

```python
#Sort employee by salary from higest to lowest
df.sort_values(by="Salary",ascending=False)
```

Sort by Department and Salary

```python
#First sort by department within each department sort salaries in descending order
df.sort_values(by=["Department","Salary"], ascending=[True,False])
```

Step 4 : Perform Aggeregation

Calculate Total Salary

```python
# Calculate the total salary
df["Salary"].sum()
```

Calculate Average salary

```python
#Calculate Average salary
df["Salary"].mean()
```

Find highest salary

```python
#Display maxium salary
df["Salary"].max()
```

Find Lowest Salary

```python
#Display the minimum salary
df["Salary"].min()
```

Count Employees

```python
# Count the total numbers of employees
df["Emp_ID"].count()
```

Step 5 : Perform GroupBy Operation

Average Salary department-wise

```python
# Group employees based on department
df.groupby("Department")["Salary"].mean()
```

Minimum Salary Department-wise

```python
#Minimum Salary Department-wise
df.groupby("Department")["Salary"].min()
```

Count Salary Department-wise

```python
#Count Salary Department-wise
df.groupby("Department")["Salary"].count()
```

Total salary Department-wise

```python
# Calculate the total salary paid by each department
df.groupby("Department")["Salary"].sum()
```

Maximum salary department-wise

```python
#Maximum salary department-wise
df.groupby("Department")["Salary"].max()
```

Average salary City-wise

```python
#Average salary City-wise
df.groupby("City")["Salary"].mean()
```

Step 6: Multiple Aggregation Function

```python
#Multiple Aggregation Function
df.groupby("Department")["Salary"].agg(["count","sum","min","max"])
```

Step 7 : Save the data

```python
# Save the data
df.to_csv("Employee_data.csv",index=False)
print("Data saved sucessfully.")
```

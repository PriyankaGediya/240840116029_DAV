Step 1: Import the pandas library

```python
#Pandas is used to create, read, modify, and analyze tabular data.
import pandas as pd
import matplotlib.pyplot as plt
```

Step 2: Create a sample data

```python
# Create a common dataset
Data = {
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    "Product": ["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet"],
    "Region": ["North","South","East","West","North","South","East","West","North","South","East","West"],
    "Sales": [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Units_Sold": [12,30,25,14,35,30,18,38,32,21,45,40]
}
```

```python
#Convert the dictionary into a Pandas DataFrame.
df = pd.DataFrame(Data)
df
```

Step 3: Create a Line Chart

```python
#The company's sales change from January to December?
plt.figure(figsize=(10, 5))

plt.plot(Data["Month"], Data["Sales"], marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

Interpretation: Line Chart – Monthly Sales

The line chart shows how sales change from January to December. Sales generally increase throughout the year, from ₹1,20,000 in January to ₹2,50,000 in December. There are small decreases in April and August.

Step 4: Create a Bar Chart

```python
#Product generated how much total sales
plt.figure(figsize=(10, 5))

plt.bar(Data["Product"], Data["Sales"])

plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

Interpretation: Bar Chart – Total Sales by Product

The bar chart compares total sales of different products. Tablet has the highest sales, followed by Mobile, while Laptop has the lowest sales.

Step 5: Create a Histogram

```python
#Are the individual sales values distributed
plt.figure(figsize=(10, 5))

plt.hist(Data["Sales"], bins=10, edgecolor="black")

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

Interpretation: Histogram

The histogram shows the distribution of sales values. Most sales values are concentrated toward the higher range, showing that the company's sales generally increased over time.

Step 6: Create a Scatter Plot

```python
#There a relationship betweeen the number of units sold and total sales
plt.figure(figsize=(10, 5))

plt.scatter(Data["Units_Sold"], Data["Sales"])

plt.title("Sales vs Units Sold")
plt.xlabel("Units Sold")
plt.ylabel("Sales")

plt.show()
```

Interpretation: Scatter Plot – Units Sold vs Sales

The scatter plot shows the relationship between the number of units sold and total sales. The points generally move upward, which indicates a positive relationship. As the number of units sold increases, total sales also tend to increase.

Step 7: Create a Pie Chart

```python
#Percentage of total sales comes from each region
plt.figure(figsize=(10, 5))

region_sales = df.groupby("Region")["Sales"].sum()

plt.pie(region_sales, labels=region_sales.index, autopct="%1.1f%%", startangle=90)

plt.title("Sales Distribution by Region")

plt.show()
```

Interpretation: Pie Chart – Sales by Region

The pie chart shows the percentage contribution of each region to total sales.
North: 22.2%
South: 24.6%
East: 26.5%
West: 26.7%
West contributes the highest percentage of total sales, while North contributes the lowest.

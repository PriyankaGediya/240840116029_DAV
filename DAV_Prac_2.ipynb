Practical - 2 : Perform data inspection, cleaning, handling missing values, duplicate removal and data transformation using pandas.

 Step : 1 import pandas

```python
#import pandas
import pandas as pd
```

Step : 2 Create a dataset

```python
#add data
data = {
    "ID":[101,102,103,104,104,105,106,107],
    "Name":["Rahul","Priya","Amit",None,"Riya","Ajay","Naresh","Jenil"],
    "Age":[20,21,None,20,20,22,23,28],
    "City":["Surat","Navsari","Surat","Bharuch","Bharuch","Navsari","Baroda","Ahmedabad"],
    "Marks":[85,None,78,88,88,81,90,93]
}
```

```python
df = pd.DataFrame(data)
df
```

Step : 3 Inspect the dataset

```python
#Display first five row of dataset.
df.head()
```

```python
#display last five row of dataset
df.tail()
```

```python
# display detail information about dataset
df.info()
```

```python
#generate statistical information for numerical colums
df.describe()
```

```python
#display dimention of dataset
#output format( Number of row, Number of column)
df.shape
```

```python
#display the name of all columns in the dataframe
df.columns
```

```python
#display the data type of each column
df.dtypes
```

Step 4 : Check missing values

```python
#check for missing value in each column
df.isnull().sum()
```

Step 5 : Replace missing values

```python
#replace missing values in the marks column
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df
```

```python
#replace missing values in the age column
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

```python
#replace missing names with the next "unknown"
df["Name"] = df["Name"].fillna("unknown")
df
```

Step 6 : Check whether

```python
#check whether each show is duplicated.
df.duplicated()
```

```python
#count total number of duplicate rows.
df.duplicated().sum()
```

Step 7 : Remove duplicate records.

```python
#Remove duplicate rows from the data frame.
df = df.drop_duplicates()
df
```

Step 8: Rename a column

```python
#Rename the column "Marks" to "Score"
df.rename(columns = {"Marks" : "Score"}, inplace = True)
df
```

Step 9 : Change data type.

```python
#Convert the age column from float to integer.
df["Age"]=df["Age"].astype(int)
df.dtypes
```

Step 10 : Create a new column.

```python
#Create a new column "Final Score"
df["Final Score"] = df["Score"] = 5
df
```

Step 11 : Convert a student name

```python
#Convert all student names to Uppercase.
df["Name"] = df["Name"].str.upper()
df
```

Step 12 : Convert text to lowercase

```python
# Convert City names to lowercase
df["City"] = df["City"].str.lower()
df
```

Step 13 : Remove extra spaces

```python
# Remove extra spaces from Name column
df["City"] = df["City"].str.strip()
df
```

Step 14 : sort the dataset

```python
#sort the dataframe based on the "Final Score" column.
df.sort_values(by = "Final Score", ascending = False)
```

Step 15 : Save the cleaned dataset

```python
#Save the cleaned dataframe as a new CSV file
df.to_csv("Cleaned_Students.csv",index=False)
print("Dataset Saved Successfully.")
```
